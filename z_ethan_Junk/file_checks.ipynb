```python
import pandas as pd
from pathlib import Path

# Read CSV into DataFrame
csv_path = Path("c:\\Users\\ethan\\code\\research\\ReEDS-2.0\\runs\\test1_nuclear_stor_small\\inputs_case\\financials_tech.csv")

df_financials_tech = pd.read_csv(csv_path)

# Check for duplicates across all columns
duplicate_mask = df_financials_tech.loc[:, ['i', 'country', 't']].duplicated(keep=False)
#get duplicates
duplicates = df_financials_tech[duplicate_mask]
duplicate_counts = duplicates.groupby(['i', 'country', 't']).size().reset_index(name='duplicate_count')
duplicate_counts
```
